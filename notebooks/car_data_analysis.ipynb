```python
# 1. Fill 'KMs Driven' with mean
df['KMs Driven'] = pd.to_numeric(df['KMs Driven'], errors='coerce')
df['KMs Driven'] = df['KMs Driven'].fillna(df['KMs Driven'].mean())

# 2. Fill all other columns with mode
for col in df.columns:
    if col != 'KMs Driven':
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python

df['KMs Driven'] = pd.to_numeric(df['KMs Driven'], errors='coerce')
df['KMs Driven'] = df['KMs Driven'].fillna(df['KMs Driven'].mean())

# 2. Fill all other columns with their mode
mode_cols = ['Brand', 'Condition', 'Fuel', 'Model', 'Registered City', 'Transaction Type', 'Year']

for col in mode_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```
