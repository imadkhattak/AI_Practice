```python
import pandas as pd
from matplotlib.pyplot import matplotlib as plt
```

```python
df = pd.read_csv(r"C:\Users\SART\Desktop\Practise\csvs\drone-strikes-dataset.csv")
df.head()
```

```python
df.info()
```

```python
import matplotlib.pyplot as plt

# 1. Group by Year and sum the 'No of Strikes'
strikes_by_year = df.groupby('Year')['No of Strikes'].sum().dropna()

# 2. Create the line chart
plt.figure(figsize=(10, 5))
plt.plot(
    strikes_by_year.index,
    strikes_by_year.values,
    marker="o",
    color="#1f77b4",
    linewidth=2,
    linestyle="-",
)

# 3. Add titles and labels
plt.title("Number of Drone Strikes by Year", fontsize=14, fontweight="bold")
plt.xlabel("Year", fontsize=12)
plt.ylabel("Total Number of Strikes", fontsize=12)

# Ensure x-axis ticks show integer years
plt.xticks(strikes_by_year.index.astype(int), rotation=45)
plt.grid(True, linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```
