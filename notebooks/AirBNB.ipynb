```python
missing_cols = df.isnull().sum()
missing_cols = missing_cols[missing_cols > 0].index

num_null_cols = df[missing_cols].select_dtypes(include='number').columns

df_numeric_nulls = df[num_null_cols]
df_numeric_nulls.head()
```

```python
for col in df_numeric_nulls.columns:
    mean_val = df[col].mean()
    df[col] = df[col].fillna(mean_val)  


cols = df_numeric_nulls.columns
df[cols] = df[cols].fillna(df[cols].mean())


df_numeric_nulls = df[cols]


df_numeric_nulls.head()

```

```python
missing_cols = df.isnull().sum()
missing_cols = missing_cols[missing_cols > 0].index

num_null_cols = df[missing_cols].select_dtypes(include='object').columns

df_categorical_nulls = df[num_null_cols]
df_categorical_nulls.head()
```

```python
df_categorical_nulls.isnull().sum()
```

```python
for col in df_categorical_nulls.columns:
    mean_val = df[col].mode()
    df[col] = df[col].fillna(mean_val)  


cols = df_categorical_nulls.columns
df[cols] = df[cols].fillna(df[cols].mode())


df_categorical_nulls = df[cols]


df_categorical_nulls.head()

```

```python
df_categorical_nulls.isnull().sum()
```

```python
cols = df_categorical_nulls.columns

# Use .iloc[0] to take the first row of modes as a series mapping column -> value
df[cols] = df[cols].fillna(df[cols].mode().iloc[0])

# Refresh df_categorical_nulls
df_categorical_nulls = df[cols]
df_categorical_nulls.head()
```

```python
df_categorical_nulls.isnull().sum()
```

```python
# Update df in-place with the imputed values
df[df_numeric_nulls.columns] = df_numeric_nulls
df[df_categorical_nulls.columns] = df_categorical_nulls
```

```python
import ast

# 1. Fill nulls with empty list representation
df['host_verifications'] = df['host_verifications'].fillna("[]")

# 2. Extract Feature 1: Count of verifications
# (Safely converts string list to actual list and takes the length)
df['verifications_count'] = df['host_verifications'].apply(
    lambda x: len(ast.literal_eval(x)) if isinstance(x, str) and x.startswith('[') else 0
)

# 3. Extract Feature 2: Binary flags for the most common verification types
df['verif_phone'] = df['host_verifications'].str.contains('phone', na=False).astype(int)
df['verif_email'] = df['host_verifications'].str.contains('email', na=False).astype(int)
df['verif_work_email'] = df['host_verifications'].str.contains('work_email', na=False).astype(int)

# 4. Drop the raw column (no longer needed)
df.drop(columns=['host_verifications'], inplace=True)
```

```python
df[['verif_phone', 'verif_email', 'verif_work_email']].tail()
```

```python
df['bathrooms_text'].value_counts()
```

```python
# 1. Convert text to lowercase and replace "half-bath" variations with 0.5
baths_cleaned = df['bathrooms_text'].str.lower().replace({
    'half-bath': '0.5',
    'shared half-bath': '0.5',
    'private half-bath': '0.5'
})

# 2. Extract digits and decimals (e.g., 1, 1.5, 2.5) and convert to float
df['bathrooms'] = baths_cleaned.str.extract(r'(\d+\.?\d*)')[0].astype(float)

# (Optional) Drop the old text column
df.drop(columns=['bathrooms_text'], inplace=True)
```
